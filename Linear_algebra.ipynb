***`What is t-SNE ?`***

0_vz4K-fqiy19u_W-9.webp

*t-SNE (t-Distributed Stochastic Neighbor Embedding) is a technique to:

Reduce high-dimensional data  2D or 3D map for visualization
Mainly for visualization*

Hold on to those who are close to you, and try to keep those who are far away away.

*Math or Algorithm based on Probability + Distance + Optimization, Used as an Algorithm*

Close points means High probability , Far Points low probability .

Original data = complex 100D Space , t-SNE = rearranging points in 2D.


PCA Ask which direction is best and t-SNE asks, who us close to whom ?

***`Example`***

`Suppose 3 cluster in 100D:`
`Dogs , Cats, Cars . t-SNE will show, 3 separate cluster in 2D, even if original space is complex.`

*So Basically PCA is a global structure and t-SNE is a local structure.*

# ***`What is Norm L1 & Norm L2 `***

``*Both L1 and L2 are ways to measure (length) or error.*```

***`If i have a Vector (List of Number) `***
``X = [x1, x2, x3]``

``How big is this vector ?``

``There are 2 populat ways ``

``L1 (Manhattan Distance)``

``L1 = |X1| + |X2| + |X3| . Just add the values``

``e.g., X = 3, 7, -9``

``L1 = |3| + |7| + |-9| = 3 + 7 + 9 = 19 . ``

``Because distance can't be  minus (-) ,  Imagine you are walking in a city like kolkata, you can't go diagonally, only . Up / Down, left / right , So distance step by step movement, that is L1``
  

Screenshot 2026-04-10 at 6.51.56 AM.png

***`What is L2 ? `***

``Think of L2 (Euclidean) distance as the "As the Crow Flies" measurement.
``
``L2 = root X1 square + X2 square + X3 square``
``x = [3, -4, 2]``

``L2 = root 3 square + -4 square + 2 sqaure = root 9 + 16 + 4 = root 29 = 5.38516481 ``

``Now you can go straight (diagonal). shortest path between two points``

Screenshot 2026-04-10 at 6.52.43 AM.png

***`WHy Use this ? `***

``L1 (Lasso Regression)``

``Forces some values -> 0, use for features selections, Remove Useless features``

***`L2 (Ridge Regression)`***

``Makes values small but not 0, keeps all features, In short the meaning is that, Reduce impact, don't remove``

Screenshot 2026-04-26 at 10.15.12 AM.png

Screenshot 2026-04-10 at 6.58.06 AM.png

````python
import numpy as np  # numerical computation er jonno

v1 = np.array([1, 2, 3])  # vector 1
v2 = np.array([4, 5, 6])  # vector 2

print(np.linalg.norm(v1, ord=1))
#L1 norm - Lasso regression (Sparsity)

print(np.linalg.norm(v1, ord=2))
#L2 norm - Ridge regression, distance calculation
````

***K-NN is a simple ML algorithm used for , Classification and Regression***

``Near by points are similar, look at k nearest data points Decide on them``

``Suppose i want to classify a new point: i have already label data, 🔴 Red is Class A and 🔵 Blue is Class B, Now a new points comes``

***Where used?

Pattern recognition
Basic classification problems

# **`Matrices (Basic operation)`**

````python
A = np.array([[1, 2], [3, 4]])  # matrix A
B = np.array([[5, 6], [7, 8]])  # matrix B

print(A + B)
# matrix addition - combine transformations

print(np.dot(A, B))
# matrix multiplication - neural networks, linear transformation

print(A.T)
# transpose - data reshape, covariance calculation

print(np.linalg.inv(A))
# inverse - solving linear regression equation

print(np.eye(2))
# identity matrix - neutral transformation
````

# ***`System of linear equation`***

````python
A = np.array([[2, 1], [1, 3]])  # coefficients, 2x, 1x or 1y , 3y = 2x + 1y = 8 , 1x + 3y = 13
B = np.array([8, 13])           # outputs

solution = np.linalg.solve(A, B)
print(solution)

# solving equations -  linear regression parameter solving
````
