# ***`Functions (Single and Multivariable) Coding`***

***`Single variable function (One input)`***

````python
import numpy as np

# f(x) = x^2 + 2x + 1
def f(x):
    return x**2 + 2*x + 1

x = 3
y = f(x)

print(y)
# single variable function → one input (used in simple models)
````

***`Multi variable function`***

````python
import numpy as np

# f(x, y) = x^2 + y^2 + 2xy
def f(x, y):
    return x**2 + y**2 + 2*x*y

x = 2
y = 3

result = f(x, y)

print(result)
# multivariable function → multiple inputs (real ML models)
````

````python
import numpy as np

# f(x1, x2) = w1*x1 + w2*x2 + b
def model(x1, x2):
    w1, w2 = 0.5, 1.5  # weights
    b = 1               # bias
    return w1*x1 + w2*x2 + b

output = model(2, 3)

print(output)
# multivariable function → linear regression / neural network
````

# ***`power, product, quotient, chain rule coding`***

***`Power rule`***

````python
import sympy as sp  # symbolic math

x = sp.symbols('x')

f = x**3  # f(x) = x^3

derivative = sp.diff(f, x)

print(derivative)
# power rule → d/dx (x^n) = n*x^(n-1)
````

*`Product rule`*

````python
import sympy as sp

x = sp.symbols('x')

f = x**2 * sp.sin(x)  # f(x) = x^2 * sin(x)

derivative = sp.diff(f, x)

print(derivative)
# product rule → (u*v)' = u'v + uv'
````

***`Quotient rule`***

````python
import sympy as sp

x = sp.symbols('x')

f = (x**2) / (x + 1)  # f(x) = x^2 / (x+1)

derivative = sp.diff(f, x)

print(derivative)
# quotient rule → (u/v)' = (u'v - uv') / v^2
````

***`Chain Rule`***

````python
import sympy as sp

x = sp.symbols('x')

f = (x**2 + 1)**3  # f(x) = (g(x))^3

derivative = sp.diff(f, x)

print(derivative)
# chain rule → derivative of outer * derivative of inner
````

# ***`Partial Derivatives coding`***

***`Multi variable function`***

````python
import sympy as sp  # symbolic math

# define variables
x, y = sp.symbols('x y')

# function: f(x, y) = x^2 + y^2 + 2xy
f = x**2 + y**2 + 2*x*y

# partial derivative w.r.t x
df_dx = sp.diff(f, x)

# partial derivative w.r.t y
df_dy = sp.diff(f, y)

print("df/dx:", df_dx)
print("df/dy:", df_dy)

# partial derivatives → ekta variable change, onno gula constant dhora hoy
````

***`Evaluate at a point`***

````python
# substitute values (x=2, y=3)

val_dx = df_dx.subs({x: 2, y: 3})
val_dy = df_dy.subs({x: 2, y: 3})

print("df/dx at (2,3):", val_dx)
print("df/dy at (2,3):", val_dy)

# ML → gradient (slope in each direction)
````

# ***`What is Gradient Vector ?`***

# ***`Gradient Vector Coding`***

***`Gradient multivariable function`***

````python
import sympy as sp  # symbolic math

# define variables
x, y = sp.symbols('x y')

# function: f(x, y) = x^2 + y^2 + 2xy
f = x**2 + y**2 + 2*x*y

# partial derivatives
df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)

# gradient vector
gradient = [df_dx, df_dy]

print("Gradient:", gradient)
# gradient vector → [∂f/∂x, ∂f/∂y] (direction of steepest increase)
````

***`Evaluate gradient at a point`***

````python
# substitute values (x=2, y=3)

grad_x = df_dx.subs({x: 2, y: 3})
grad_y = df_dy.subs({x: 2, y: 3})

print("Gradient at (2,3):", [grad_x, grad_y])
# ML → slope in each direction (used in gradient descent)
````

# ***`What is Directional Derivatives`***

# ***`Directional Derivatives coding`***

# *** Directional Derivative (Multivariable Function)***

````python
import sympy as sp  # symbolic math

# define variables
x, y = sp.symbols('x y')

# function: f(x, y)
f = x**2 + y**2

# gradient (∇f)
grad_f = [sp.diff(f, x), sp.diff(f, y)]

print("Gradient:", grad_f)
# gradient → direction of maximum increase
````

***` Directional Derivative in a Given Direction`***

````python
# direction vector (e.g., v = [1, 1])
v = sp.Matrix([1, 1])

# normalize direction vector (unit vector)
v_unit = v / sp.sqrt(v.dot(v))

# convert gradient to vector
grad_vec = sp.Matrix(grad_f)

# directional derivative = ∇f ⋅ v_unit
directional_derivative = grad_vec.dot(v_unit)

print("Directional Derivative:", directional_derivative)
# directional derivative → specific direction e rate of change
````

***`Evaluate at a point`***

````python
# evaluate at (x=1, y=2)
value = directional_derivative.subs({x: 1, y: 2})

print("Value at (1,2):", value)
# ML → direction wise change (optimization path)
````

# ***`Higher-order derivatives and Taylor series (first-order approximation is enough) coding`***

***`Higher-Order Derivatives`***

````python
import sympy as sp  # symbolic math

x = sp.symbols('x')

# function: f(x) = x^3 + 2x^2 + x
f = x**3 + 2*x**2 + x

# first derivative
f1 = sp.diff(f, x)

# second derivative
f2 = sp.diff(f, x, 2)

# third derivative
f3 = sp.diff(f, x, 3)

print("f'(x):", f1)
print("f''(x):", f2)
print("f'''(x):", f3)
# higher-order derivatives → curvature, acceleration of change
````

***`Taylor series (First order approximation)`***

````python
import sympy as sp

x = sp.symbols('x')

# function
f = sp.sin(x)

# point a = 0
a = 0

# first-order Taylor approximation: f(a) + f'(a)(x - a)
f_a = f.subs(x, a)
f_prime = sp.diff(f, x)
f_prime_a = f_prime.subs(x, a)

taylor_approx = f_a + f_prime_a * (x - a)

print("Taylor Approximation:", taylor_approx)
# first-order Taylor → linear approximation near point
````

***`Evaluate Approximation`***

````python
# compare real vs approximation at x = 0.1

real = sp.sin(0.1)
approx = taylor_approx.subs(x, 0.1)

print("Real value:", real)
print("Approx value:", approx)
# approximation → nearby value estimate
````

# ***`Optimization in One variable in One coding`***

***`Optimization in One variable (Find Minimum / Maximum)`***

````python
import sympy as sp  # symbolic math

x = sp.symbols('x')

# function: f(x) = x^2 - 4x + 4
f = x**2 - 4*x + 4

# first derivative
f_prime = sp.diff(f, x)

# find critical points (set derivative = 0)
critical_points = sp.solve(f_prime, x)

print("Critical Point:", critical_points)
# optimization → jekhane slope = 0 (possible min/max)
````

***`Check Minimum or Maximum (Second Derivative)`***

````python
# second derivative
f_double_prime = sp.diff(f, x, 2)

for point in critical_points:
    value = f_double_prime.subs(x, point)

    if value > 0:
        print(f"Minimum at x = {point}")
    elif value < 0:
        print(f"Maximum at x = {point}")
````

***`Find Minimum Value`***

````python
# evaluate function at critical point
min_value = f.subs(x, critical_points[0])

print("Minimum value:", min_value)
# ML → loss function minimize kora (same concept)
````

# ***`Critical points, first and second derivative test coding`***

***`Critical Points (Where f(x)=0)`***

````python
import sympy as sp  # symbolic math

x = sp.symbols('x')

# function: f(x) = x^3 - 3x^2 + 2
f = x**3 - 3*x**2 + 2

# first derivative
f_prime = sp.diff(f, x)

# find critical points
critical_points = sp.solve(f_prime, x)

print("Critical Points:", critical_points)
# critical points → slope = 0 (possible min/max)
````

***`First Derivative test`***

````python
for point in critical_points:
    left = f_prime.subs(x, point - 0.1)   # just left of point
    right = f_prime.subs(x, point + 0.1)  # just right of point

    if left > 0 and right < 0:
        print(f"Maximum at x = {point}")
    elif left < 0 and right > 0:
        print(f"Minimum at x = {point}")
    else:
        print(f"No change at x = {point}")
# first derivative test → slope change check
````

***`Second derivative test`***

````python
# second derivative
f_double_prime = sp.diff(f, x, 2)

for point in critical_points:
    value = f_double_prime.subs(x, point)

    if value > 0:
        print(f"Minimum at x = {point}")
    elif value < 0:
        print(f"Maximum at x = {point}")
    else:
        print(f"Inconclusive at x = {point}")
# second derivative test → curvature check
````

# ***`What is Local Vs GLobal minima ?`***

# ***`Local vs Global minima coding`***

````python
import numpy as np
import matplotlib.pyplot as plt

# define function (multiple minima)
def f(x):
    return x**4 - 4*x**2 + x  # complex curve

# generate values
x = np.linspace(-3, 3, 100)
y = f(x)

# plot
plt.plot(x, y)
plt.axhline(0)
plt.axvline(0)
plt.title("Local vs Global Minima")
plt.show()
# visualize → multiple low points exist
````

***`Find Critical Points`***

````python
import sympy as sp

x = sp.symbols('x')

f = x**4 - 4*x**2 + x

# first derivative
f_prime = sp.diff(f, x)

# critical points
critical_points = sp.solve(f_prime, x)

print("Critical Points:", critical_points)
# candidates for local/global minima
````

***`Check Minima Values`***

````python
# evaluate function at critical points
values = [(point, f.subs(x, point)) for point in critical_points]

print("Points and Values:", values)

# global minimum = lowest value
global_min = min(values, key=lambda x: x[1])

print("Global Minimum:", global_min)
# local minima → nearby lowest
# global minima → overall lowest
````

# ***`Convex function Coding`***

````python
import numpy as np
import matplotlib.pyplot as plt

# convex function: f(x) = x^2
def f(x):
    return x**2

# generate values
x = np.linspace(-5, 5, 100)
y = f(x)

# plot
plt.plot(x, y)
plt.title("Convex Function: f(x) = x^2")
plt.show()
# convex → bowl shape (one global minimum)
````

***` Check Convexity (Second Derivative > 0)`***

````python
import sympy as sp

x = sp.symbols('x')

f = x**2  # convex function

# second derivative
f_double_prime = sp.diff(f, x, 2)

print("Second Derivative:", f_double_prime)
# if f''(x) > 0 → convex
````

***`Non-Convex Example`***

````python
import numpy as np
import matplotlib.pyplot as plt

# non-convex function
def g(x):
    return x**4 - 4*x**2

x = np.linspace(-3, 3, 100)
y = g(x)

plt.plot(x, y)
plt.title("Non-Convex Function")
plt.show()
# non-convex → multiple minima (hard optimization)
````

# ***`Coding`***

***`Indefinite Integral`***

````python
import sympy as sp  # symbolic math

x = sp.symbols('x')

# function: f(x) = 2x
f = 2*x

# indefinite integral
integral = sp.integrate(f, x)

print(integral)
# ∫2x dx = x^2 + C
# indefinite → general form (area function)
````

***`Definite Integral`***

````python
import sympy as sp

x = sp.symbols('x')

# function
f = 2*x

# definite integral from 0 to 2
result = sp.integrate(f, (x, 0, 2))

print(result)
# definite → exact area under curve (0 to 2)
````

***`Area Under Curve (Visualization)`***

````python
import numpy as np
import matplotlib.pyplot as plt

# function
def f(x):
    return 2*x

x = np.linspace(0, 2, 100)
y = f(x)

plt.plot(x, y)
plt.fill_between(x, y, alpha=0.3)
plt.title("Area Under Curve")
plt.show()
# shaded area → integral (total accumulation)
````
