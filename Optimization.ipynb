#***`Objective  / Loss function ? (e.g., MSE , cross-entrophy) Coding`***

***`MSE`***

```python
import numpy as np  # numerical operations

# actual values
y_true = np.array([3, 5, 2, 7])

# predicted values
y_pred = np.array([2.5, 5, 4, 8])

# MSE formula
mse = np.mean((y_true - y_pred) ** 2)

print(mse)
# MSE → squared error (regression problems)


```

***`Cross-Entrophy`***

```python
import numpy as np

# actual labels (0 or 1)
y_true = np.array([1, 0, 1, 1])

# predicted probabilities
y_pred = np.array([0.9, 0.2, 0.8, 0.7])

# avoid log(0)
epsilon = 1e-10
y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

# binary cross-entropy
loss = -np.mean(
    y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred)
)

print(loss)
# cross-entropy → probability-based loss (classification)
```

# ***`Cost function (Convex vs Non convex) coding`***

***`Convex Cost function`***

```python
import numpy as np
import matplotlib.pyplot as plt

# convex cost function (MSE type)
def convex_loss(x):
    return x**2  # simple bowl shape

x = np.linspace(-5, 5, 100)
y = convex_loss(x)

plt.plot(x, y)
plt.title("Convex Cost Function")
plt.show()
# convex → one global minimum (easy optimization)
```

***`Non-Convex Cost Function`***

```python
import numpy as np
import matplotlib.pyplot as plt

# non-convex function
def non_convex_loss(x):
    return x**4 - 4*x**2 + x  # multiple minima

x = np.linspace(-3, 3, 100)
y = non_convex_loss(x)

plt.plot(x, y)
plt.title("Non-Convex Cost Function")
plt.show()
# non-convex → multiple local minima (hard optimization)
```

***`Compare Both Together`***

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 100)

convex = x**2
non_convex = x**4 - 4*x**2 + x

plt.plot(x, convex, label="Convex")
plt.plot(x, non_convex, label="Non-Convex")

plt.legend()
plt.title("Convex vs Non-Convex")
plt.show()
# comparison → convex easy, non-convex tricky
```

# ***`Global vs Local minima saddle points Coding`***

```python
import numpy as np
import matplotlib.pyplot as plt

# function with multiple minima
def f(x):
    return x**4 - 4*x**2 + x

x = np.linspace(-3, 3, 100)
y = f(x)

plt.plot(x, y)
plt.title("Local vs Global Minima")
plt.show()
# multiple low points → local & global minima
```

***`Find critical points`***

```python
import sympy as sp

x = sp.symbols('x')

f = x**4 - 4*x**2 + x

# first derivative
f_prime = sp.diff(f, x)

# critical points
critical_points = sp.solve(f_prime, x)

print("Critical Points:", critical_points)
# candidates for minima/maxima/saddle
```

***`Check Type (Min / max)`***

```python
# second derivative
f_double_prime = sp.diff(f, x, 2)

for point in critical_points:
    value = f_double_prime.subs(x, point)

    if value > 0:
        print(f"Local Minimum at x = {point}")
    elif value < 0:
        print(f"Local Maximum at x = {point}")
    else:
        print(f"Saddle Point at x = {point}")
```

***`Find Global minima`***

```python
values = [(point, f.subs(x, point)) for point in critical_points]

global_min = min(values, key=lambda x: x[1])

print("Global Minimum:", global_min)
# lowest among all → global minimum
```

#***`Gradient Descent Coding`***

***`Gradient Descent basic coding`***

```python
import numpy as np

# function: f(x) = x^2 (want to minimize)
def f(x):
    return x**2

# derivative: f'(x) = 2x
def grad(x):
    return 2*x

# initial value
x = 5

# learning rate
lr = 0.1

# iterations
for i in range(20):
    x = x - lr * grad(x)  # gradient descent step

print("Minimum at x:", x)
# gradient descent → step by step minimum e jawa
```

# ***`What is Batch gradient Descent ?`***

Screenshot 2026-04-13 at 3.40.19 PM.png

# ***`Batch Gradient Descent coding`***

```python
import numpy as np  # numerical operations

# sample data (X → feature, y → target)
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# parameters (initial)
w = 0.0  # weight
b = 0.0  # bias

lr = 0.01  # learning rate
epochs = 100  # iterations

n = len(X)

for i in range(epochs):

    # predictions
    y_pred = w * X + b

    # gradients (using ALL data → batch)
    dw = (-2/n) * np.sum(X * (y - y_pred))
    db = (-2/n) * np.sum(y - y_pred)

    # update parameters
    w = w - lr * dw
    b = b - lr * db

print("Weight:", w)
print("Bias:", b)
# batch gradient descent → full dataset use kore update
```

#***`Stochastic Gradient Descent Coding`***

```python
import numpy as np  # numerical operations

# sample data
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# parameters
w = 0.0  # weight
b = 0.0  # bias

lr = 0.01
epochs = 50

n = len(X)

for epoch in range(epochs):
    for i in range(n):

        xi = X[i]      # single data point
        yi = y[i]

        y_pred = w * xi + b

        # gradients (only one sample)
        dw = -2 * xi * (yi - y_pred)
        db = -2 * (yi - y_pred)

        # update
        w = w - lr * dw
        b = b - lr * db

print("Weight:", w)
print("Bias:", b)
# SGD → one data point use kore update (fast but noisy)
```

# ***`What is Mini batch GD`***

# ***`Mini batch GD Coding`***

```python
import numpy as np  # numerical operations

# sample data
X = np.array([1, 2, 3, 4, 5, 6])
y = np.array([2, 4, 6, 8, 10, 12])

# parameters
w = 0.0  # weight
b = 0.0  # bias

lr = 0.01
epochs = 50
batch_size = 2  # small batch

n = len(X)

for epoch in range(epochs):

    # shuffle data (important)
    indices = np.random.permutation(n)
    X_shuffled = X[indices]
    y_shuffled = y[indices]

    # mini-batch loop
    for i in range(0, n, batch_size):
        X_batch = X_shuffled[i:i+batch_size]
        y_batch = y_shuffled[i:i+batch_size]

        # predictions
        y_pred = w * X_batch + b

        # gradients (batch er upor)
        dw = (-2/len(X_batch)) * np.sum(X_batch * (y_batch - y_pred))
        db = (-2/len(X_batch)) * np.sum(y_batch - y_pred)

        # update
        w = w - lr * dw
        b = b - lr * db

print("Weight:", w)
print("Bias:", b)
# mini-batch → small group data diye update (balanced approach)
```

# ***`Momentum and accelerated methods (Nesterov) Coding`***

***`Gradient Descent with Momentum`***

```python
import numpy as np  # numerical operations

# function: f(x) = x^2
def grad(x):
    return 2*x  # derivative

x = 5          # initial value
lr = 0.1       # learning rate
momentum = 0.9 # momentum factor
v = 0          # velocity (initial)

for i in range(20):
    g = grad(x)             # gradient
    v = momentum * v + lr * g  # update velocity
    x = x - v               # update position

    print(f"Step {i+1}: x = {x}")
# momentum → smooth update, faster convergence
```

***`Nesterov accelerated gradient (NAG)`***

```python
import numpy as np

# function: f(x) = x^2
def grad(x):
    return 2*x

x = 5
lr = 0.1
momentum = 0.9
v = 0

for i in range(20):
    # lookahead position
    x_lookahead = x - momentum * v

    g = grad(x_lookahead)      # gradient at lookahead
    v = momentum * v + lr * g  # update velocity
    x = x - v                  # update position

    print(f"Step {i+1}: x = {x}")
# Nesterov → future position dekhe update (more accurate)
```

#***`What is Adaptive optimizers: Adam, RMSprop `***

***`RMSprop (Adaptive Learning Rate)`***

```python
import numpy as np  # numerical operations

# function: f(x) = x^2
def grad(x):
    return 2*x

x = 5
lr = 0.1
beta = 0.9   # decay rate
epsilon = 1e-8

s = 0  # running average of squared gradients

for i in range(20):
    g = grad(x)

    s = beta * s + (1 - beta) * (g**2)  # update squared gradient avg

    x = x - (lr / (np.sqrt(s) + epsilon)) * g  # adaptive update

    print(f"Step {i+1}: x = {x}")
# RMSprop → learning rate automatically adjust hoy
```

***`Adam optimizer (Momentum + RMSprop)`***

```python
import numpy as np

# function: f(x) = x^2
def grad(x):
    return 2*x

x = 5
lr = 0.1

beta1 = 0.9  # momentum term
beta2 = 0.999  # RMS term
epsilon = 1e-8

m = 0  # first moment (mean)
v = 0  # second moment (variance)

for t in range(1, 21):
    g = grad(x)

    # update moments
    m = beta1 * m + (1 - beta1) * g
    v = beta2 * v + (1 - beta2) * (g**2)

    # bias correction
    m_hat = m / (1 - beta1**t)
    v_hat = v / (1 - beta2**t)

    # update
    x = x - (lr * m_hat) / (np.sqrt(v_hat) + epsilon)

    print(f"Step {t}: x = {x}")
# Adam → most popular optimizer (fast + stable)
```

***`L1 Regularization (Lasso)`***

```python
import numpy as np  # numerical operations

# weights
w = np.array([1.5, -2.0, 0.5])

# L1 norm (sum of absolute values)
l1 = np.sum(np.abs(w))

print(l1)
# L1 → |w1| + |w2| + ...
# use → feature selection (some weights → 0)
```

***`L2 Regularization (Ridge)`***

```python
import numpy as np

# weights
w = np.array([1.5, -2.0, 0.5])

# L2 norm (sum of squared values)
l2 = np.sum(w**2)

print(l2)
# L2 → w1^2 + w2^2 + ...
# use → reduce large weights (no zero, smooth)
```

***`L1 vs L2 in Loss function`***

```python
import numpy as np

# true and predicted
y_true = np.array([3, 5, 2])
y_pred = np.array([2.5, 5, 4])

# weights
w = np.array([1.5, -2.0, 0.5])

# MSE loss
mse = np.mean((y_true - y_pred) ** 2)

# add regularization
lambda_ = 0.1

l1_loss = mse + lambda_ * np.sum(np.abs(w))
l2_loss = mse + lambda_ * np.sum(w**2)

print("L1 Loss:", l1_loss)
print("L2 Loss:", l2_loss)
# regularization → overfitting control
```

# ***`Early stoping and Dropout Coding `***

***`Early stopping (Stop when validation stops improving)`***

```python
import numpy as np

# dummy training & validation loss
train_loss = [1.0, 0.8, 0.6, 0.5, 0.4, 0.35, 0.34]
val_loss   = [1.1, 0.9, 0.7, 0.6, 0.65, 0.7, 0.75]

patience = 2  # how many epochs to wait
best_loss = float("inf")
counter = 0

for epoch in range(len(train_loss)):

    print(f"Epoch {epoch+1}, Val Loss: {val_loss[epoch]}")

    if val_loss[epoch] < best_loss:
        best_loss = val_loss[epoch]
        counter = 0  # reset
    else:
        counter += 1

    if counter >= patience:
        print("Early Stopping Triggered")
        break

# early stopping → overfitting hole training stop kore
```

***`Dropout (Randomly drop neurons)`***

```python
import numpy as np

# input layer (example)
X = np.array([1.0, 2.0, 3.0, 4.0])

dropout_rate = 0.5  # 50% neurons drop

# create mask (0 or 1)
mask = np.random.binomial(1, 1 - dropout_rate, size=X.shape)

# apply dropout
X_dropped = X * mask

print("Original:", X)
print("Mask:", mask)
print("After Dropout:", X_dropped)
# dropout → random neuron off kore overfitting komay
```
