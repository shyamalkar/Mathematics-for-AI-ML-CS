`The chain rule is used to find the probability of multiple events happening together step-by-step.   So the Basic Idea is that :- Instead of calculating everything at once, we break it into parts.`

# ***`Conditional Probability (P(A | B))`***

# P(A and B), P(B)
P_A_and_B = 0.2
P_B = 0.5

# Conditional probability: P(A | B)
P_A_given_B = P_A_and_B / P_B

print(P_A_given_B)
# conditional probability - What is the chance if event B already occuring and B occured .

## ***`Chain Rule (Multiple Events)`***

````python
# P(A), P(B | A), P(C | A and B)
P_A = 0.6
P_B_given_A = 0.5
P_C_given_AB = 0.4

# Chain rule: P(A and B and C)
P_ABC = P_A * P_B_given_A * P_C_given_AB

print(P_ABC)
# chain rule → multiple dependent events er joint probability
````

# ***`Independence (Independent event) Coding`***

````python
# P(A), P(B)
P_A = 0.5
P_B = 0.4

# For independent events:
# P(A and B) = P(A) * P(B)
P_A_and_B = P_A * P_B

print(P_A_and_B)
# independence → When one event never depend on another one.
````

# ***`Mutual Exclusivity`***

````python
# P(A), P(B)
P_A = 0.3
P_B = 0.4

# For mutually exclusive events:
# P(A and B) = 0
P_A_and_B = 0

# Addition rule simplifies:
P_A_or_B = P_A + P_B

print(P_A_or_B)
# mutual exclusivity → duita event ekshathe ghotte pare na
````

# ***`Bayes theorem coding`***

````python
# P(A), P(B|A), P(B)
P_A = 0.6        # prior - A howar chance
P_B_given_A = 0.7  # likelihood - B given A
P_B = 0.5        # evidence - B howar total probability

# Bayes Theorem: P(A | B)
P_A_given_B = (P_B_given_A * P_A) / P_B

print(P_A_given_B)
# Bayes - updated probability after new evidence (B)
````

# ***`Permutations`***

````python
import math  # factorial use korar jonno

n = 5  # total items
r = 2  # select koto ta

# Permutation formula: nPr = n! / (n-r)!
perm = math.factorial(n) / math.factorial(n - r)

print(perm)
# permutation → order important (ranking, arrangement)
````

# ***`Combinations`***

````python
import math  # factorial use korar jonno

n = 5  # total items
r = 2  # select koto ta

# Combination formula: nCr = n! / (r! * (n-r)!)
comb = math.factorial(n) / (math.factorial(r) * math.factorial(n - r))

print(comb)
# combination → order important na (selection only)
````

#***`Covariance`***

````python
import numpy as np  # numerical operations

# two variables (features)
X = np.array([1, 2, 3, 4, 5])
Y = np.array([2, 4, 6, 8, 10])

cov_matrix = np.cov(X, Y)

print(cov_matrix)
# covariance → duita variable ekshathe change kore kina (direction)
# +ve → same direction, -ve → opposite direction
````

# ***`Correlation`***

````python
import numpy as np

# same data
X = np.array([1, 2, 3, 4, 5])
Y = np.array([2, 4, 6, 8, 10])

corr_matrix = np.corrcoef(X, Y)

print(corr_matrix)
# correlation → relationship strength (-1 to 1)
# 1 → strong positive, -1 → strong negative, 0 → no relation
````

#***`What is Expectation mean, variance, std ? `***

Screenshot 2026-04-11 at 8.07.27 PM.png

# ***`Coding`***

````python
import statistics

# Test scores for a class
scores = [85, 90, 78, 92, 88, 70, 95]

# 1. Expectation (Mean)
avg = statistics.mean(scores)

# 2. Variance (The spread)
var = statistics.variance(scores)

# 3. Standard Deviation (The measurement)
stdev = statistics.stdev(scores)

print(f"Typical Score (Mean): {avg:.2f}")
print(f"How messy is the pile? (Variance): {var:.2f}")
print(f"Average distance from center: {stdev:.2f}")

# ML Logic -  Is a score of 40 'weird'?
if 40 < (avg - (2 * stdev)):
    print("Warning: This score is an Outlier (too far from the mean)!")
````

# ***`PMF, PDF, CDF - Coding`***

#**PMF (Probability Mass Function )**

````python
import numpy as np

# discrete data (dice outcomes)
values = np.array([1, 2, 3, 4, 5, 6])

# PMF → each value er probability (equal for fair dice)
pmf = np.ones(len(values)) / len(values)

print(pmf)
# PMF → discrete values er exact probability (classification, count data)
````

# ***`Probability Density Function`***

````python
import numpy as np
from scipy.stats import norm  # normal distribution

x = np.linspace(-3, 3, 5)  # range of values

pdf = norm.pdf(x, loc=0, scale=1)

print(pdf)
# PDF → continuous data er density (exact probability na, area important)
````

# ***`What is CDF (Comulative Distribution Function)`***

````python
import numpy as np
from scipy.stats import norm

x = np.linspace(-3, 3, 5)

cdf = norm.cdf(x, loc=0, scale=1)

print(cdf)
# CDF → x porjonto total probability (<= x)
````
